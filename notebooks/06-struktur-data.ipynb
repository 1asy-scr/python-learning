# Struktur Data: List, Tuple, Dictionary

Struktur data menyimpan sekumpulan nilai.

## 1. List

```python
buah = ['apel', 'jeruk', 'mangga']
print(buah)
buah.append('pisang')
print('Setelah ditambah:', buah)
print('Elemen pertama:', buah[0])
```

## 2. Tuple
Tuple mirip list, tetapi elemennya tidak dapat diubah setelah dibuat.

```python
koordinat = (10, 20)
print(koordinat)
print('x =', koordinat[0])
```

## 3. Dictionary

```python
mahasiswa = {'nama': 'Bima', 'prodi': 'Pendidikan Matematika', 'semester': 2}
print(mahasiswa['nama'])
mahasiswa['semester'] = 3
print(mahasiswa)
```
