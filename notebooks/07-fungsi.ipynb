# Fungsi dalam Python

Fungsi adalah blok kode yang dapat digunakan kembali.

## 1. Fungsi sederhana

```python
def sapa():
    print('Selamat belajar Python!')

sapa()
```

## 2. Parameter dan return

```python
def luas_persegi_panjang(panjang, lebar):
    return panjang * lebar

hasil = luas_persegi_panjang(8, 5)
print('Luas:', hasil)
```

## 3. Latihan
Buat fungsi untuk mengubah Celsius menjadi Fahrenheit. Rumus: F = (C × 9/5) + 32.

```python
def celsius_ke_fahrenheit(celsius):
    return (celsius * 9/5) + 32

print(celsius_ke_fahrenheit(30))
```
