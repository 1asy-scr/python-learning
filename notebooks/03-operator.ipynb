# Operator Python

Kita akan mencoba operator aritmetika, perbandingan, dan logika.

## 1. Operator aritmetika

```python
a = 17
b = 5
print('Tambah:', a + b)
print('Kurang:', a - b)
print('Kali:', a * b)
print('Bagi:', a / b)
print('Sisa bagi:', a % b)
print('Pangkat:', a ** b)
```

## 2. Operator perbandingan

```python
x = 10
y = 12
print(x == y)
print(x < y)
print(x != y)
```

## 3. Operator logika

```python
umur = 20
punya_kartu = True
print(umur >= 17 and punya_kartu)
print(umur < 17 or punya_kartu)
```
