# Perulangan

Perulangan membantu menjalankan instruksi berkali-kali.

## 1. Perulangan for

```python
for i in range(1, 6):
    print('Perulangan ke-', i)
```

## 2. Perulangan while

```python
hitung = 1
while hitung <= 5:
    print(hitung)
    hitung += 1
```

## 3. Akumulasi
Hitung jumlah bilangan 1 sampai 100.

```python
total = 0
for angka in range(1, 101):
    total += angka
print('Jumlah:', total)
```
