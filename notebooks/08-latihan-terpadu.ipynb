## Tantangan 1: Kalkulator sederhana

```python
a = 15
b = 3
print('Penjumlahan:', a + b)
print('Pengurangan:', a - b)
print('Perkalian:', a * b)
print('Pembagian:', a / b)
```

## Tantangan 2: Menentukan kategori usia

```python
usia = 18
if usia < 13:
    kategori = 'Anak-anak'
elif usia < 18:
    kategori = 'Remaja'
else:
    kategori = 'Dewasa'
print(kategori)
```

## Tantangan 3: Konversi suhu

```python
def celsius_ke_fahrenheit(c):
    return c * 9/5 + 32

for c in [0, 25, 37, 100]:
    print(c, 'C =', celsius_ke_fahrenheit(c), 'F')
```
