# Variabel dan Tipe Data

Variabel digunakan untuk menyimpan nilai.

## 1. Membuat variabel

```python
nama = 'Alya'
umur = 19
berat = 48.5
mahasiswa = True

print(nama)
print(umur)
print(berat)
print(mahasiswa)
```

## 2. Memeriksa tipe data
Gunakan `type()` untuk mengetahui tipe suatu nilai.

```python
print(type(nama))
print(type(umur))
print(type(berat))
print(type(mahasiswa))
```
