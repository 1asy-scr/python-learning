# Percabangan

Percabangan menjalankan instruksi berdasarkan kondisi.

## 1. If dan else

```python
nilai = 78
if nilai >= 70:
    print('Lulus')
else:
    print('Belum lulus')
```

## 2. If, elif, dan else

```python
nilai = 85
if nilai >= 90:
    predikat = 'A'
elif nilai >= 80:
    predikat = 'B'
elif nilai >= 70:
    predikat = 'C'
else:
    predikat = 'Perlu perbaikan'
print('Predikat:', predikat)
```

## 3. Latihan
Buat program untuk menentukan bilangan positif, negatif, atau nol.

```python
angka = -4
if angka > 0:
    print('Positif')
elif angka < 0:
    print('Negatif')
else:
    print('Nol')
```
